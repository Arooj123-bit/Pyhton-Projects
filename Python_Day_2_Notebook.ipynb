## while loop

```python
password = 123
user_entry = int(input("Please enter your password:"))

while password != user_entry:
  print("Password is incorrect. Try again.")
  user_entry = int(input("Please enter your password:"))
  if password == user_entry:
    print("Access is granted.")
    break
```

```python
print("Lets play a number guessing game.")
print("I picked a number. Try to guess it.")
secret_number = 20
guess = int(input("Guess the number:"))

if secret_number == guess:
  print("You won!. You are a genius")
else:
  while secret_number != guess:
    print("Incorrect try again.")
    guess = int(input("Guess the number:"))
    if secret_number == guess:
       print("You won!. You are a genius")
```
